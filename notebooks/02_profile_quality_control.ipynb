```python
from pathlib import Path
import pandas as pd

RAW = Path("../data/raw")

meta = pd.read_csv(RAW / "quality_control_meta.csv",encoding='cp1251')
structure = pd.read_csv(RAW / "quality_control_structure.csv", encoding='cp1251')

meta
```

```python
DATA_PATH = RAW / "quality_control_data.csv"

df = pd.read_csv(
    DATA_PATH,
    sep=";",
    encoding="cp1251",
    dtype=str,
    keep_default_na=False,
)

print(df.shape)
df.head(10)
```

```python
df.columns.tolist()
```

```python
for col in df.columns:
    print("\n" + "=" * 60)
    print(col)
    print("Уникальных значений:", df[col].nunique())
    print("Пустых значений:", (df[col] == "").sum())
    print(df[col].value_counts(dropna=False).head(10))
```

```python
DATE_COL = "letter_date"

df[DATE_COL] = pd.to_datetime(df[DATE_COL], errors="coerce", dayfirst=True)

print("Минимальная дата:", df[DATE_COL].min())
print("Максимальная дата:", df[DATE_COL].max())
print("Уникальных месяцев:", df[DATE_COL].dt.to_period("M").nunique())
print("Строк с нераспознанной датой:", df[DATE_COL].isna().sum())
```

```python
df[DATE_COL].dt.to_period("M").value_counts().sort_index().plot(
    kind="bar",
    figsize=(14, 4),
    title="Количество записей по месяцам",
)
```

```python
quality_report = pd.DataFrame({
    "column": df.columns,
    "dtype": [df[c].dtype for c in df.columns],
    "non_empty": [(df[c] != "").sum() for c in df.columns],
    "empty": [(df[c] == "").sum() for c in df.columns],
    "unique": [df[c].nunique() for c in df.columns],
})

quality_report["empty_share"] = quality_report["empty"] / len(df)
quality_report.sort_values("empty_share", ascending=False)
```

```python
quality_report.to_csv("../data/processed/quality_report.csv", index=False)
```
